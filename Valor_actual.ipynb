# CALCULAR VALOR PRESENTE

```python
import pandas as pd

# con este codigo aplicaremos el concepto de valor actual
def valor_actual(cuotas, interes_anual):

    i = interes_anual / 12 / 100

    vp = 0

    for t, cuota in enumerate(cuotas, start=1):

        vp += cuota / ((1 + i) ** t) 

    return vp 
```

```python

# calculo del sistema frances de amortización teniendo en cuenta como se calcula la cuota fija mensual
def cuotas_frances(capital, interes_anual, n):

    i = interes_anual / 12 / 100

    cuota = (
        capital *
        i *
        (1 + i) ** n
    ) / (
        (1 + i) ** n - 1
    )

    return [cuota] * n



# calculo del sistema aleman de amortización teniendo en cuenta como se calcula la cuota fija mensual
def cuotas_aleman(capital, interes_anual, n):

    i = interes_anual / 12 / 100

    saldo = capital

    amortizacion = capital / n

    cuotas = []

    for _ in range(n):

        intereses = saldo * i

        cuota = amortizacion + intereses

        cuotas.append(cuota)

        saldo -= amortizacion

    return cuotas



#El valor actual a mitad del prestamo
def valor_actual_mitad(cuotas, interes_anual):

    mitad = len(cuotas) // 2

    cuotas_restantes = cuotas[mitad:]

    i = interes_anual / 12 / 100

    VA = 0

    for t, Mt in enumerate(cuotas_restantes, start=1):

        VA += Mt / ((1 + i) ** t)

    return VA


```

```python
df = pd.read_csv(
    "data/clientes_clasificados_frances_aleman.csv"
)

resultado = []

for _, fila in df.iterrows():

    capital = fila["Monto_Inicial"]
    interes = fila["Ratio_Interes"]
    duracion = int(fila["Duracion"])

    if fila["Sistema_Recomendado"] == "Francés":

        cuotas = cuotas_frances(
            capital,
            interes,
            duracion
        )

    else:

        cuotas = cuotas_aleman(
            capital,
            interes,
            duracion
        )

    va = valor_actual_mitad(
        cuotas,
        interes
    )

    resultado.append(va)

df["Valor_Actual_Mitad"] = resultado

df.to_csv(
    "clientes_con_valor_actual.csv",
    index=False
)
# Para saber que se ha terminado de ejecutar el codigo y que se ha generado el archivo "clientes_con_valor_actual.csv"
print("Terminado")
```
