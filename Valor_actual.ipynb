```python
import pandas as pd

df = pd.read_csv("data/clientes_clasificados_frances_aleman.csv", encoding="utf-8-sig")

# Interés mensual y número de cuotas
df["i"] = df["Ratio_Interes"] / 100 / 12
df["n"] = df["Duracion"]

# Factor de actualización (valor actual de 1 euro al mes durante n meses)
df["a_n"] = (1 - (1 + df["i"]) ** -df["n"]) / df["i"]

# --- Sistema francés: renta constante, temporal, vencida ---
df["Cuota_Fr"] = df["Monto_Inicial"] / df["a_n"]
df["VA_Fr"] = df["Cuota_Fr"] * df["a_n"]
df["VF_Fr"] = df["VA_Fr"] * (1 + df["i"]) ** df["n"]

# --- Sistema alemán: renta variable (progresión aritmética), temporal, vencida ---
df["Amortizacion_Al"] = df["Monto_Inicial"] / df["n"]
df["C1_Al"] = df["Amortizacion_Al"] + df["Monto_Inicial"] * df["i"]
df["d_Al"] = -df["Amortizacion_Al"] * df["i"]
df["VA_Al"] = ((df["C1_Al"] + df["d_Al"] / df["i"]) * df["a_n"]
               - (df["n"] * df["d_Al"] / df["i"]) * (1 + df["i"]) ** -df["n"])
df["VF_Al"] = df["VA_Al"] * (1 + df["i"]) ** df["n"]
```

```python
print((df["VA_Fr"] - df["Monto_Inicial"]).abs().max())
print((df["VA_Al"] - df["Monto_Inicial"]).abs().max())
```
